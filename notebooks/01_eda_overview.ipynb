```python
import pandas as pd
import matplotlib.pyplot as plt
import getpass

from sqlalchemy import create_engine

print("Ambiente carregado com sucesso!")
```

```python
query = """
SELECT
    s.season,
    m.match_date,

    home.name AS home_team,
    away.name AS away_team,

    m.home_goals,
    m.away_goals,
    m.full_time_result,

    m.home_half_time_goals,
    m.away_half_time_goals,
    m.half_time_result,

    m.referee,

    m.home_shots,
    m.away_shots,

    m.home_shots_on_target,
    m.away_shots_on_target,

    m.home_fouls,
    m.away_fouls,

    m.home_corners,
    m.away_corners,

    m.home_yellow_cards,
    m.away_yellow_cards,

    m.home_red_cards,
    m.away_red_cards

FROM matches m

JOIN seasons s
    ON s.id = m.season_id

JOIN teams home
    ON home.id = m.home_team_id

JOIN teams away
    ON away.id = m.away_team_id

ORDER BY
    m.match_date,
    m.id;
"""
```

```python
senha = getpass.getpass("Digite a senha do PostgreSQL: ")

engine = create_engine(
    f"postgresql+psycopg://postgres:{senha}@127.0.0.1:5432/premier_league_analytics"
)

print("Engine PostgreSQL criada com sucesso!")
```

```python
df = pd.read_sql_query(query, engine)

print(f"Linhas: {df.shape[0]}")
print(f"Colunas: {df.shape[1]}")
```

```python
df.head()
```

```python
# Criar coluna com o total de gols de cada partida
df["total_goals"] = df["home_goals"] + df["away_goals"]

# Resumo por temporada
gols_por_temporada = (
    df.groupby("season")
    .agg(
        partidas=("season", "size"),
        total_gols=("total_goals", "sum"),
        media_gols_por_jogo=("total_goals", "mean")
    )
    .reset_index()
)

# Arredondar a média
gols_por_temporada["media_gols_por_jogo"] = (
    gols_por_temporada["media_gols_por_jogo"].round(2)
)

gols_por_temporada
```

```python
plt.figure(figsize=(11, 5))

plt.plot(
    gols_por_temporada["season"],
    gols_por_temporada["media_gols_por_jogo"],
    marker="o",
    linewidth=2
)

plt.title(
    "Evolução da média de gols por partida na Premier League",
    fontsize=14
)

plt.xlabel("Temporada")
plt.ylabel("Média de gols por partida")

plt.xticks(rotation=45)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()
```

```python
# Visão das equipes como mandantes
mandantes = pd.DataFrame({
    "season": df["season"],
    "match_date": df["match_date"],
    "team": df["home_team"],
    "opponent": df["away_team"],
    "venue": "home",
    "goals_for": df["home_goals"],
    "goals_against": df["away_goals"],
    "result": df["full_time_result"].map({
        "H": "W",
        "D": "D",
        "A": "L"
    })
})


# Visão das equipes como visitantes
visitantes = pd.DataFrame({
    "season": df["season"],
    "match_date": df["match_date"],
    "team": df["away_team"],
    "opponent": df["home_team"],
    "venue": "away",
    "goals_for": df["away_goals"],
    "goals_against": df["home_goals"],
    "result": df["full_time_result"].map({
        "H": "L",
        "D": "D",
        "A": "W"
    })
})


# Unir as duas perspectivas
df_clubes = pd.concat(
    [mandantes, visitantes],
    ignore_index=True
)


# Calcular pontos
df_clubes["points"] = df_clubes["result"].map({
    "W": 3,
    "D": 1,
    "L": 0
})


df_clubes.head()
```

```python
print(f"Linhas: {df_clubes.shape[0]}")
print(f"Colunas: {df_clubes.shape[1]}")
```

```python
ranking_clubes = (
    df_clubes.groupby("team")
    .agg(
        jogos=("team", "size"),
        vitorias=("result", lambda x: (x == "W").sum()),
        empates=("result", lambda x: (x == "D").sum()),
        derrotas=("result", lambda x: (x == "L").sum()),
        gols_pro=("goals_for", "sum"),
        gols_contra=("goals_against", "sum"),
        pontos=("points", "sum"),
        temporadas=("season", "nunique")
    )
    .reset_index()
)

ranking_clubes["saldo_gols"] = (
    ranking_clubes["gols_pro"]
    - ranking_clubes["gols_contra"]
)

ranking_clubes["aproveitamento_pct"] = (
    ranking_clubes["pontos"]
    / (ranking_clubes["jogos"] * 3)
    * 100
).round(2)

ranking_clubes = ranking_clubes.sort_values(
    by="pontos",
    ascending=False
)

ranking_clubes.head(15)
```

```python
desempenho_temporada = (
    df_clubes
    .groupby(["season", "team"])
    .agg(
        jogos=("team", "size"),
        vitorias=("result", lambda x: (x == "W").sum()),
        empates=("result", lambda x: (x == "D").sum()),
        derrotas=("result", lambda x: (x == "L").sum()),
        gols_pro=("goals_for", "sum"),
        gols_contra=("goals_against", "sum"),
        pontos=("points", "sum")
    )
    .reset_index()
)

# Saldo de gols
desempenho_temporada["saldo_gols"] = (
    desempenho_temporada["gols_pro"]
    - desempenho_temporada["gols_contra"]
)

# Pontos por jogo
desempenho_temporada["pontos_por_jogo"] = (
    desempenho_temporada["pontos"]
    / desempenho_temporada["jogos"]
).round(2)

# Aproveitamento
desempenho_temporada["aproveitamento_pct"] = (
    desempenho_temporada["pontos"]
    / (desempenho_temporada["jogos"] * 3)
    * 100
).round(2)

desempenho_temporada.head(20)
```

```python
print(f"Linhas: {desempenho_temporada.shape[0]}")
print(f"Colunas: {desempenho_temporada.shape[1]}")
```

```python
consistencia_clubes = (
    desempenho_temporada
    .groupby("team")
    .agg(
        temporadas=("season", "nunique"),
        media_pontos_jogo=("pontos_por_jogo", "mean"),
        desvio_pontos_jogo=("pontos_por_jogo", "std"),
        minimo_pontos_jogo=("pontos_por_jogo", "min"),
        maximo_pontos_jogo=("pontos_por_jogo", "max"),
        media_aproveitamento=("aproveitamento_pct", "mean")
    )
    .reset_index()
)

# Diferença entre a melhor e a pior temporada
consistencia_clubes["amplitude_pontos_jogo"] = (
    consistencia_clubes["maximo_pontos_jogo"]
    - consistencia_clubes["minimo_pontos_jogo"]
)

# Arredondar
colunas_arredondar = [
    "media_pontos_jogo",
    "desvio_pontos_jogo",
    "minimo_pontos_jogo",
    "maximo_pontos_jogo",
    "media_aproveitamento",
    "amplitude_pontos_jogo"
]

consistencia_clubes[colunas_arredondar] = (
    consistencia_clubes[colunas_arredondar]
    .round(2)
)

# Considerar apenas clubes com pelo menos 5 temporadas
consistencia_clubes = consistencia_clubes[
    consistencia_clubes["temporadas"] >= 5
]

consistencia_clubes.sort_values(
    by="media_pontos_jogo",
    ascending=False
).head(15)
```

```python
plt.figure(figsize=(12, 7))

plt.scatter(
    consistencia_clubes["media_pontos_jogo"],
    consistencia_clubes["desvio_pontos_jogo"],
    s=80
)

# Adicionar nome dos clubes
for _, clube in consistencia_clubes.iterrows():
    plt.annotate(
        clube["team"],
        (
            clube["media_pontos_jogo"],
            clube["desvio_pontos_jogo"]
        ),
        xytext=(5, 5),
        textcoords="offset points",
        fontsize=8
    )


# Linhas de referência usando as medianas
mediana_media = (
    consistencia_clubes["media_pontos_jogo"].median()
)

mediana_desvio = (
    consistencia_clubes["desvio_pontos_jogo"].median()
)


plt.axvline(
    mediana_media,
    linestyle="--",
    linewidth=1
)

plt.axhline(
    mediana_desvio,
    linestyle="--",
    linewidth=1
)


plt.title(
    "Desempenho x Consistência dos clubes na Premier League",
    fontsize=14
)

plt.xlabel("Média de pontos por jogo")
plt.ylabel("Desvio padrão dos pontos por jogo")

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
def classificar_clube(linha):

    if (
        linha["media_pontos_jogo"] >= mediana_media
        and linha["desvio_pontos_jogo"] <= mediana_desvio
    ):
        return "Acima da mediana + consistente"

    elif (
        linha["media_pontos_jogo"] >= mediana_media
        and linha["desvio_pontos_jogo"] > mediana_desvio
    ):
        return "Acima da mediana + volátil"

    elif (
        linha["media_pontos_jogo"] < mediana_media
        and linha["desvio_pontos_jogo"] <= mediana_desvio
    ):
        return "Abaixo da mediana + consistente"

    else:
        return "Abaixo da mediana + volátil"


consistencia_clubes["perfil"] = (
    consistencia_clubes.apply(
        classificar_clube,
        axis=1
    )
)

consistencia_clubes[
    [
        "team",
        "media_pontos_jogo",
        "desvio_pontos_jogo",
        "perfil"
    ]
].sort_values(
    by="media_pontos_jogo",
    ascending=False
)
```

```python
# Z-score da média de pontos por jogo
consistencia_clubes["z_desempenho"] = (
    consistencia_clubes["media_pontos_jogo"]
    - consistencia_clubes["media_pontos_jogo"].mean()
) / consistencia_clubes["media_pontos_jogo"].std()


# Z-score da variabilidade
consistencia_clubes["z_variabilidade"] = (
    consistencia_clubes["desvio_pontos_jogo"]
    - consistencia_clubes["desvio_pontos_jogo"].mean()
) / consistencia_clubes["desvio_pontos_jogo"].std()
```

```python
consistencia_clubes["idc_bruto"] = (
    consistencia_clubes["z_desempenho"]
    - consistencia_clubes["z_variabilidade"]
)
```

```python
idc_min = consistencia_clubes["idc_bruto"].min()
idc_max = consistencia_clubes["idc_bruto"].max()

consistencia_clubes["idc"] = (
    (
        consistencia_clubes["idc_bruto"] - idc_min
    )
    /
    (
        idc_max - idc_min
    )
    * 100
).round(2)
```

```python
### Índice de Desempenho Consistente (IDC)

O IDC foi desenvolvido neste projeto como uma métrica exploratória para
combinar desempenho esportivo e estabilidade entre temporadas.

O desempenho é representado pela média de pontos por jogo, enquanto a
consistência é obtida a partir da variabilidade desse indicador entre
temporadas.

Ambas as métricas são normalizadas em uma escala de 0 a 100.

O índice é calculado por:

**IDC = Score de Desempenho × (0,70 + 0,30 × Score de Consistência / 100)**

Dessa forma, o desempenho constitui a base do indicador, enquanto a
consistência funciona como fator de ajuste. Assim, um clube de baixo
desempenho não obtém uma classificação elevada apenas por apresentar
baixa variabilidade.

O IDC é uma métrica exploratória criada para este estudo e não corresponde
a um indicador oficial da Premier League.
```
